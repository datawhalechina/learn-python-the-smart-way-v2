# 欢迎来到面向对象编程的世界！

当程序逻辑变得复杂时，仅使用独立的变量和函数会让代码难以维护。

**面向对象编程 (OOP)** 是一种将 **数据（属性）** 与 **操作数据的行为（方法）** 打包封装在一起的编程范式。

在本章中，我们将以 **“扑克牌”** 为例，循序渐进地学习 OOP 的基础知识。


# 方法与函数

在设计具体的扑克牌对象之前，我们先来观察一个在 Python 中常用的调用结构差异。


### 主谓结构与动宾结构

在之前的学习中，我们见过这两种调用方式：
1. `len(s)` —— **函数 (Function)**：动宾结构（动作 + 宾语），如 `测量(字符串)`。
2. `s.upper()` —— **方法 (Method)**：主谓结构（主语 + 动作），如 `字符串.变大写()`。

下面用代码展示两者的区别：


````python
# 假设我们有一张牌的名字字符串
card_name = 'Hearts Ace' # 红桃 A
````

函数写法 (动宾结构)：我们把 `card_name` 传入内置函数 `len()`，测量名字的长度。

````python
print(len(card_name))
````

方法写法 (主谓结构)：`card_name` 自己调用 `upper()` 方法，把自己变成大写。

````python
print(card_name.upper())
````

替换内容的方法：`card_name` 自己调用 `replace()` 方法。

````python
print(card_name.replace('Hearts', 'Spades'))
````

调用出错时的报错信息也不同：


函数调用错误测试。

````python
number = 1
print(len(number))
````

方法调用错误测试。

````python
print(number.upper())
````

# 类和实例

- **类 (Class)**：对象的模板或图纸。例如：定义所有扑克牌都有花色和点数的“扑克牌模板”。
- **实例 (Instance)**：根据模板创建的具体对象。例如：一张真正的“红桃 A”纸牌。

在 Python 中：
- `int`, `str` 是**类**（类型）。
- `10`, `"Hearts"` 是对应类生成的**实例**（具体的值）。


# 一切皆对象

在 Python 中，**一切皆对象**：
- 每一个具体的值都是对象。
- 每一个对象都有它所属的类。
- 类本身也是一个类型为 `type` 的对象。

OOP 的核心思想：以“对象”为单位组织代码，将属性和方法封装在一起。

# 编写构造函数

每次手动给新对象赋值属性（如 `c1.suit = ...`）很繁琐。

**构造函数 `__init__`** 可以在创建对象时，自动进行属性的初始化设置。

我们希望以后能像这样，在卡牌创建时就传入花色和点数：
```python
c1 = Card('红桃', 1)
```

如果让我们自己设计初始化函数，可能写成这样：
```python
def constructor(card, suit, number):
    card.suit = suit
    card.number = number
```

Python 官方约定使用双下划线包裹的特殊方法 `__init__` 作为构造函数：

它的基本结构是这样的：
```python
def __init__(card, suit, number):
    card.suit = suit
    card.number = number
```

规范要求将第一个参数（代表对象自己）统一命名为 `self`：
```python
def __init__(self, suit, number):
    self.suit = suit      # 把你传进来的 suit 赋给“自己(self)”的 suit 属性
    self.number = number  # 把你传进来的 number 赋给“自己(self)”的 number 属性
```

完整的类初始化实现如下：


````python
class Card:
    def __init__(self, suit, number):
        self.suit = suit
        self.number = number
 
c1 = Card('红桃', 1)
c2 = Card('黑桃', 13)

print(f"初始化成功！c1是 {c1.suit}{c1.number}，c2是 {c2.suit}{c2.number}")
````

# 编写方法

在类中定义的函数称为**方法 (Method)**，代表该类对象可以执行的行为。

我们通过以下步骤，将一个外部函数改写为类的方法：


````python
class Card:
    def __init__(self, suit, number):
        self.suit = suit
        self.number = number
````

这是一个普通的外部函数（不是类的方法），它需要接收一个 `card` 对象作为参数：

````python
def show(card):
    print(f'牌面是: {card.suit} {card.number}')

c1 = Card('红桃', 1)
c2 = Card('黑桃', 13)

show(c1) 
show(c2)
````

把函数移入类中，变成方法。

我们将函数移到类 `Card` 内部（缩进），并改为主谓结构的调用方式：


````python
class Card:
    def __init__(self, suit, number):
        self.suit = suit
        self.number = number

    # 缩进写在类内部，它就成了方法
    def show(card):
        print(f'牌面是: {card.suit} {card.number}')

c1 = Card('红桃', 1)
c2 = Card('黑桃', 13)

# 此时 Python 会自动把 c1 作为第一个参数传给 show 方法中的 card 参数。
c1.show() 
c2.show()
````

把 `card` 替换为 `self`。

将第一个参数名规范化为 `self`：

````python
class Card:
    def __init__(self, suit, number):
        self.suit = suit
        self.number = number

    # 使用规范名称 self
    def show(self):
        print(f'牌面是: {self.suit} {self.number}')

c1 = Card('红桃', 1)
c2 = Card('黑桃', 13)

c1.show()
c2.show()
````

### 方法也可以接收额外的参数：


````python
class Card:
    def __init__(self, suit, number):
        self.suit = suit
        self.number = number

    def compare_number(self, other):
        # 比较点数大小
        if self.number > other.number:
            print(f'{self.suit}{self.number} 大于 {other.suit}{other.number}')
        elif self.number < other.number:
            print(f'{self.suit}{self.number} 小于 {other.suit}{other.number}')
        else:
            print(f'{self.suit}{self.number} 与 {other.suit}{other.number} 点数相同')

c1 = Card('红桃', 10)
c2 = Card('黑桃', 1)

c1.compare_number(c2)
````

### 方法还可以修改和维护对象自身的状态（属性）：


````python
class Card:
    def __init__(self, suit, number):
        self.suit = suit
        self.number = number
        self.is_face_up = False # 默认牌面朝下

    def flip(self):
        # 翻牌：将正反面状态取反
        self.is_face_up = not self.is_face_up
        state = "正面朝上" if self.is_face_up else "背面朝上"
        print(f'翻牌啦！当前牌状态: {state}')

c = Card('红桃', 1)
c.flip() # 翻到正面
c.flip() # 翻到背面
````

# 类和方法的优点

1. **封装 (Encapsulation)**：数据和行为打包，结构清晰。
2. **高内聚，防误用**：限制方法的调用范围，避免类型误用。
3. **多态 (Polymorphism)**：同名方法在不同类中具有不同的实现。


````python
class NormalCard:
    def __init__(self, number):
        self.number = number
    def get_score(self):
        # 普通牌，点数就是得分
        return self.number

class JokerCard:
    def __init__(self, is_big):
        self.is_big = is_big
    def get_score(self):
        # 大王计 50 分，小王计 40 分
        return 50 if self.is_big else 40

for card in [NormalCard(10), JokerCard(is_big=True)]:
    print("卡牌得分:", card.get_score()) # 同一个接口，不同实现！这就是多态的魅力。
````

# 玩家手牌管理器

我们将编写一个 `Hand` 类，内部用列表存储多个 `Card` 实例（即 **“对象包含对象”** 的组合设计）。

它支持摸牌、出牌、计算总点数、整理手牌（排序）以及展示手牌：

````python
class Card:
    def __init__(self, suit, number):
        self.suit = suit
        self.number = number

    def __repr__(self):
        # 转换点数为简称，方便打印
        num_name = {1: 'A', 11: 'J', 12: 'Q', 13: 'K'}.get(self.number, str(self.number))
        return f'{self.suit}{num_name}'

class Hand:
    def __init__(self, player_name):
        self.player_name = player_name
        self.cards = [] # 手牌列表，用于存放 Card 对象实例

    def add_card(self, card):
        # 向手牌中加入一张牌
        self.cards.append(card)
        print(f"{self.player_name}摸牌: {card}")

    def discard_card(self, suit, number):
        # 从手牌中丢弃特定的一张牌
        for card in self.cards:
            if card.suit == suit and card.number == number:
                self.cards.remove(card)
                print(f"{self.player_name}打出: {card}")
                return card
        print(f"手牌中找不到 {suit}{number}，无法打出！")
        return None

    def get_total_points(self):
        # 统计手牌总点数 (Ace 计 1 分，J/Q/K 计 11/12/13 分)
        return sum(card.number for card in self.cards)

    def show_hand(self):
        # 展示手牌
        print(f"{self.player_name}的手牌: {self.cards} (总计 {self.get_total_points()} 点)")

    def sort_hand(self):
        # 对手牌按点数从小到大排序
        self.cards.sort(key=lambda card: card.number)
        print("整理牌型，手牌已按点数从小到大排序。")
````

````python
# 模拟真实的牌局摸牌与出牌流程：
print("--- 牌局开始 ---")
my_hand = Hand("玩家小鲸鱼🐳")

# 连续摸了 3 张牌
my_hand.add_card(Card('红桃', 1))  # 摸红桃 A
my_hand.add_card(Card('黑桃', 13)) # 摸黑桃 K
my_hand.add_card(Card('方块', 5))  # 摸方块 5

# 展示当前手牌与总分
my_hand.show_hand()
````

````python
# 整理手牌
my_hand.sort_hand()
my_hand.show_hand()
````

````python
# 打出一张牌
my_hand.discard_card('方块', 5)
my_hand.show_hand()
````

# 类型判断

- `type(obj) == A`：**精确匹配**，必须是 `A` 类本身。
- `isinstance(obj, A)`：**宽泛匹配**，可以是 `A` 类或其子类的实例。


````python
class Card:
    ...
class JokerCard(Card):
    ...

c = Card()
jc = JokerCard()

print(type(c) == Card)
print(type(jc) == Card) # False (jc 是王牌，不是普通的基础卡牌 Card)
print(isinstance(jc, Card)) # True  (王牌也是卡牌的一种，所以 isinstance 认为是 True)
````

# 特殊方法 (魔术方法 - Magic Methods)

以双下划线开头和结尾的特殊方法（如 `__init__`）被称为 **魔术方法** 。

它们由 Python 在特定情境下自动调用（例如，`a == b` 会自动调用 `a.__eq__(b)`）。


## 1. 相等 `__eq__`
通过重写 `__eq__` 方法，定义花色和点数相同时卡牌即相等：

````python
class Card:
    def __init__(self, suit, number):
        self.suit = suit
        self.number = number

    # 当执行 c1 == other 时，Python 自动调用此方法
    def __eq__(self, other):
        return (self.suit == other.suit and self.number == other.number)

c1 = Card('红桃', 10)
c2 = Card('红桃', 2)
print(c1 == c2)
````

但是，如果我们将纸牌和一个整数 99 比较呢？

````python
print(c1 == 99)
````

更好的做法：先用 `isinstance` 判断对方类型是否相符：


````python
class Card:
    def __init__(self, suit, number):
        self.suit = suit
        self.number = number

    def __eq__(self, other):
        # 只有当对方也是 Card 且花色、点数相同时，才返回 True
        return (isinstance(other, Card) and 
                self.suit == other.suit and 
                self.number == other.number)

c1 = Card('红桃', 10)
c2 = Card('红桃', 10)
print(c1 == c2)  # 输出: True
print(c1 == 99)  # 输出: False (优雅地返回了 False，没有崩溃！)
````

## 2. 打印自己：`__str__` 和 `__repr__`

默认情况下打印自定义对象会输出内存地址。
- `__str__`：面向普通用户，控制 `print()` 和 `str()` 的友好输出。
- `__repr__`：面向开发者调试，控制列表等容器内的输出。如果未定义 `__str__`，Python 默认使用 `__repr__`。


````python
class Card:
    def __init__(self, suit, number):
        self.suit = suit
        self.number = number
c = Card('红桃', 10)
print(c)
````

定义 `__str__` 以展示友好文本：


````python
class Card:
    def __init__(self, suit, number):
        self.suit = suit
        self.number = number
    def __str__(self):
        num_name = {1: 'A', 11: 'J', 12: 'Q', 13: 'K'}.get(self.number, str(self.number))
        return f'{self.suit}{num_name}'

c = Card('红桃', 1)
print(c)
print([c]) # 可是，如果把它放在手牌列表里打印，依然显示的是原始地址格式。
````

定义 `__repr__` 以控制调试和列表中的展示：


````python
class Card:
    def __init__(self, suit, number):
        self.suit = suit
        self.number = number
    def __repr__(self):
        return f"Card('{self.suit}', {self.number})"

c = Card('红桃', 1)
print(c)   # 默认借用 repr
print([c]) # 列表里也是清晰的表达式
````

# 继承 (Inheritance) 

**子类 (Subclass)** 可以继承 **父类 (Superclass)** 的属性和方法，减少重复代码。


## 指定父类

在定义子类时，在类名圆括号中写入父类：`class SubClass(ParentClass):`。


````python
class Card:
    def __init__(self, number):
        self.number = number
    def show_number(self):    
        print(f"点数是: {self.number}")

# StandardCard 继承自 Card
class StandardCard(Card):
    def __init__(self, number, suit):
        # 继承父类的初始化过程，省去 self.number = number
        super().__init__(number)
        self.suit = suit
        
    def show_card(self):
        print(f"标准牌面: {self.suit} {self.number}")

c = Card(10)
sc = StandardCard(1, '红桃')

c.show_number()
sc.show_number()  # 子类直接继承了父类的 show_number 方法，可以直接使用！
sc.show_card()    # 子类特有的方法
````

父类不能调用子类方法。

````python
c.show_card()
````

## 重写方法
子类可以定义同名方法以覆盖（重写）父类的方法。可以使用 `super()` 调用父类原有逻辑。


````python
class Card:
    def __init__(self, number):
        self.number = number
    def show(self):
        print(f"卡牌点数: {self.number}")

class JokerCard(Card):
    def __init__(self, is_big):
        # 大王计为点数 15，小王计为点数 14
        number = 15 if is_big else 14
        super().__init__(number)
        self.is_big = is_big
        
    def show(self):
        # 重写父类的 show 方法
        role = "大王" if self.is_big else "小王"
        print(f"🃏 {role} (点数: {self.number})")

c = Card(10)
jc = JokerCard(is_big=True)

c.show()  # 普通卡牌 of show 行为
jc.show() # 王牌重写后的 show 行为
````

## 3. 继承关系中的 `isinstance` 与 `type`

- `type` 精确匹配类型。
- `isinstance` 支持继承链匹配，子类实例也会被判定为父类类型。


````python
class 鸟:
    ...
class 鸭子(鸟):
    ...

麻雀_1 = 鸟()
鸭子_1 = 鸭子()
````

````python
print(isinstance(麻雀_1, 鸟))
print(isinstance(鸭子_1, 鸟))
print(isinstance(麻雀_1, 鸭子))
````
