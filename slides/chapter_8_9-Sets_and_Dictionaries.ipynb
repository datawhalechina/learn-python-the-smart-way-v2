# 集合 (Sets)


集合（set）是一种可以无序容纳多个元素的数据结构。

我们不能通过索引访问它，但可以遍历它。

```python
s = set([2, 3, 5])
print(3 in s)
print(4 in s)
```

```python
for x in range(7):
    if x not in s:
        print(x)
```

## 创建集合


创建一个空集合。


```python
s = set()
print(s)
```

从列表创建集合。


```python
s = set(["learn", "python", "the", "smart", "way"])
print(s)
```

创建静态分配的集合（使用字面量创建）。


```python
s = {2, 3, 5}
print(s)
```

注意：`{}` 不是空集合！


```python
s = {}
print(type(s) == set)
print(type(s))
```

> 这是一个字典 `dict`，我们很快就会学到它。

## 使用集合

集合可以做许多与列表和元组相同的事情。

```python
s = set([1, 2, 3])

print(len(s))

print(2 in s)
print(4 in s)
print(4 not in s)
print(2 not in s)
```

使用 `add` 而不是 `append` 向集合中添加元素。

```python
s.add(7)
print(s)
```

从集合中移除 `3`，如果 `3` 不在 `s` 中，会抛出错误。

```python
s.remove(3)
print(s)
```

我们可以遍历 s 中的项

```python
for item in s:
    print(item)
```

## 集合的属性


虽然集合与列表和元组非常相似，但它们有一些重要的区别。


### 集合是无序的


元素的排列顺序可能与提供它们的顺序不同，
并且我们不能通过索引访问集合。

```python
s = set([2,4,8])
print(s)
```

```python
for element in s:
    print(element)
```

### 元素是唯一的


集合也只能保存每个唯一元素的一个副本。重复项会被移除。

```python
s = set([2,2,2])
print(s) 
print(len(s))
```

### 元素必须是不可变的


集合只能容纳不可变（不能更改）的元素，例如数字、布尔值、字符串和元组。

```python
a = ["lists", "are", "mutable"]
s = set([a])
print(s)
```

另一个例子：


```python
s1 = set(["sets", "are", "mutable", "too"])
s2 = set([s1])
print(s)
```

# 字典 (Dictionaries)


字典（dictionary）是一种将键（keys）映射到值（values）的数据结构，就像列表将索引映射到值一样。

但是，键必须是任何不可变的值！

```python
stateMap = {"广州": "广东", "桂林": "广西", "丹东": "辽宁"}
city = input("请输入城市名称 --> ")
if (city in stateMap):
    print(city, "在", stateMap[city])
else:
    print("抱歉，从未听说过。")
```

另一个示例：


```python
counts = {}
while True:
    n = int(input("请输入一个整数（输入 0 结束）--> "))
    if (n == 0): break
    if (n in counts):
        counts[n] += 1
    else:
        counts[n] = 1
    print("我总共看到了", n, "一共", counts[n], "次")
print("完成，计数：", counts)
```

## 创建字典


创建一个空字典。


```python
d = dict()
print(d)
```

我们也可以使用空花括号。

```python
d = {}
print(d)
```

从键值对 (key, value) 列表创建字典。


```python
pairs = [("cow", 5), ("dog", 98), ("pig", 100), ("cat", 1)]
d = dict(pairs)
print(d)    # 顺序是无法预测的（无序的）！
```

静态分配字典（直接使用字面量创建）。


```python
d = {"cow": 5, "dog": 98, "cat": 1}
print(d)
```

## 使用字典


我们可以用与列表/集合类似的方法与字典交互。

```python
d = {"a": 1, "b": 2, "c": 3}

print(len(d))

print("a" in d)
print(2 in d) # 我们检查的是键（keys），而不是值（values）
print(2 not in d)
print("a" not in d)
```

向字典添加一个新的键值对，或者更新现有键的值。

```python
d = {"a": 1, "b": 2, "c": 3 }
d["e"] = "wow"
print(d)
```

从字典中移除指定的键值对。
> 如果键不在 `d` 中，程序会崩溃。

```python
del d["e"]
print(d)
```

```python
d.pop("z", "not found")
```

查找与给定键关联的值的两种方式：

```python
d = {"a": 1, "b": 2, "c": 3}
print(d["a"]) # 如果键不在 d 中，程序会崩溃
```

如果键在字典中，则返回该键对应的值，如果键不在 `d` 中，则返回第二个（默认）值。

```python
print(d.get("z", 42))
```

遍历字典的两种方式：

```python
for key in d:
    print(key, d[key])
```

```python
for _, value in d.items():
    print(value)
```

## 字典的属性

### 字典建立键到值的映射

```python
ages = {}
key = "fred"
value = 38
ages[key] = value  # "fred" 是键，38 是对应的值
print(ages[key])
```

### 键构成一个集合


键是无序的


```python
d = {}
d[2] = 100
d[4] = 200
d[8] = 300
print(d)  # 顺序是无法预测的
```

键是唯一的


```python
d = {}
d[2] = 100
d[2] = 200
d[2] = 400
print(d)  # 输出 { 2:400 }
```

键必须是不可变的


```python
d = {}
a = [1] # 列表是可变的，因此...
d[a] = 42
```

## 值没有限制


值可以是可变的。

```python
d = {}
a = [1,2]
d["fred"] = a
print(d["fred"])
a += [3]
print(d["fred"]) # 会反映出 a 的改变！
```

但键必须是不可变的。

```python
d[a] = 42
```

# 总结
- 集合保持着数学本身的特性：无序、唯一，并且集合中的元素是不可变的（hashable）。
- 字典是键值对的集合，其键的属性与上一条相同，其值可以是任意的。
- 如果你想遍历字典，`dict.items()` 会很好用。
- 如果你不确定你的字典中是否包含某个键，可以试试 `dict.get(key)` 和 `dict.pop(key)` 的操作。
