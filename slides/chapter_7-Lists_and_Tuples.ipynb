# 创建列表

创建一个包含多个元素的列表。

```python
evens = [2, 4, 6, 8]
pets = ["cat", "dog", "pig"]
info = ["cat", "Python", 14, 15.6, True]

print(info)
print(type(info))
```

使用 `len()` 函数获取列表的长度。

```python
n = len(info)
print(n)
```

只有一个元素的列表。

```python
a = ["hello"]
b = [42]

print(type(a), len(a), a)
print(type(b), len(b), b)
```

没有元素的列表（空列表）。

```python
a = [] # faster
b = list()

print(type(a), len(a), a)
print(type(b), len(b), b)
```

## 其他创建列表的方式

创建一个包含 n 个 `0` 的新列表。

```python
n = 10
a = [0] * n
print(type(a), len(a), a)
```

使用 `list()` 把 `range()` 生成的数字序列，转换成列表。

```python
b = list(range(n))
print(type(b), len(b), b)
```

# 通过索引和切片访问元素

列表的索引和切片与字符串的工作方式相同。

```python
a = [2, 3, 5, 7, 11, 13]
print("a        =", a)

# 访问非负索引
print("a[0]     =", a[0])
print("a[2]     =", a[2])

# 访问负索引
print("a[-1]    =", a[-1])
print("a[-3]    =", a[-3])

# 访问切片 a[start:end:step]
print("a[0:2]   =", a[0:2])
print("a[1:4]   =", a[1:4])
print("a[1:6:2] =", a[1:6:2])
```

# 查找元素


检查列表成员关系：`in`。

```python
a = [2, 3, 5, 2, 6, 2, 2, 7]

print("2 in a =", (2 in a))
print("4 in a =", (4 in a))
```

检查列表非成员关系：`not in`。

```python
print("2 not in a =", (2 not in a))
print("4 not in a =", (4 not in a))
```

统计列表中出现的次数：`list.count(item)`。

```python
a = [ 2, 3, 5, 2, 6, 2, 2, 7 ]

print("a.count(1) =", a.count(1))
print("a.count(2) =", a.count(2))
print("a.count(3) =", a.count(3))
```

查找元素的索引：`list.index(item)` 和 `list.index(item, start)`。

```python
a = [2, 3, 5, 2, 6, 2, 2, 7]

print("a.index(6)   =", a.index(6))
print("a.index(2)   =", a.index(2))
print("a.index(2,1) =", a.index(2,1))
print("a.index(2,4) =", a.index(2,4))
```

错误：当元素不在列表中时会崩溃。

```python
a = [2, 3, 5, 2]

print("a.index(9) =", a.index(9)) # 崩溃！
print("这一行不会运行！")
```

解决方案：使用 `item in list` 判断。


```python
a = [2, 3, 5, 2]

if 9 in a:
    print("a.index(9) =", a.index(9))
else:
    print("9 不在", a, "中")
print("现在这一行可以运行了！")
```

# 添加元素


## 破坏性地（修改列表）


使用 `list.append(item)` 添加一个元素。

```python
a = [2, 3]

a.append(7)
print(a)
```

使用 `list += another_list` 或 `list.extend(another_list)` 添加一个列表的元素。

```python
a = [2, 3]

a += [11, 13]
print(a)

a.extend([17, 19])
print(a)
```

在给定索引处插入一个元素。

```python
a = [2, 3, 5, 7, 11]

a.insert(2, 42)  # 在索引 2 处插入 42
print(a)
```

## 非破坏性地（创建新列表）


使用 `list_1 + list_2` 添加一个元素。

```python
a = [2, 3]
b = a + [13, 17]

print(a)
print(b)
```

在给定索引处插入一个元素（使用列表切片）。

```python
a = [2, 3]
b = a[:1] + [5] + a[1:]

print(a)
print(b)
```

## 破坏性 vs 非破坏性示例


破坏性：

```python
a = [2, 3]
b = a
a += [4]

print(a)
print(b)
```

非破坏性：

```python
a = [2, 3]
b = a
a = a + [4] # 这会覆盖 a，但不会覆盖 b 的别名

print(a)
print(b)
```

# 移除元素


## 破坏性地（修改列表）


使用 `list.remove(item)` 移除一个元素。

```python
a = [2, 3, 5, 3, 7, 6, 5, 11, 13]

a.remove(5)
print("执行 a.remove(5) 后，a=", a)

a.remove(5)
print("再次执行 a.remove(5) 后，a=", a)
```

使用 `list.pop(index)` 移除给定索引处的元素。

```python
a = [2, 3, 4, 5, 6, 7, 8]

item = a.pop(3)
print("执行 item = a.pop(3) 后")
print("item =", item)
print("a =", a)
```

```python
item = a.pop(3)
print("再次执行 item = a.pop(3) 后")
print("item =", item)
print("a =", a)
```

使用 `list.pop()` 移除最后一个元素。

```python
item = a.pop()
print("执行 item = a.pop() 后")
print("item =", item)
print("a =", a)
```

## 非破坏性地（移除元素）

移除给定索引处的元素（使用列表切片）。

```python
a = [2, 3, 5, 3, 7, 5, 11, 13]

b = a[:2] + a[3:]
print("执行 b = a[:2] + a[3:] 后")
print("a =", a)
print("b =", b)
```

# 遍历列表


使用普通的 `for` 循环遍历：


```python
a = [2, 3, 5, 7]
print("这是 a 中的元素及其索引：")
for index in range(len(a)):
    print("a[", index, "] =", a[index])
```

使用 `for-each` 循环遍历：

列表和字符串都是可迭代类型，这意味着我们可以直接对它们进行迭代（循环）。

```python
a = [2, 3, 5, 7]
print("这是 a 中的元素：")
for item in a:
    print(item)
```

**危险！！！在 `for` 循环内部修改 。**


重要提示：不要在 `for` 循环内部更改列表！索引的行为将变得不可预测。


```python
a = [2, 3, 5, 3, 7]

# 尝试移除所有 3 的失败尝试
for index in range(len(a)):
    if (a[index] == 3):  # 这最终会崩溃！
        a.pop(index)

print("这一行不会运行！")
```

**同样危险：在 `for-each` 循环内部修改。**


```python
a = [3, 3, 2, 3, 4]
for item in a:       # 这不会访问到列表中的每一个元素！
    if (item == 3):
      a.remove(item)
print(a) # 应该是 [2, 4]，但里面仍然有一个 3！
```

更好的做法：在 `while` 循环内部通过手动控制索引来修改列表。

```python
a = [2, 3, 5, 3, 7]

# 成功移除所有 3 的尝试
index = 0
while (index < len(a)):
    if (a[index] == 3):
        a.pop(index)
    else:
        index += 1

print("这一行会运行！")
print(a)
```

# 列表方法：排序和反转


列表有许多内置方法，这些方法通常既有破坏性实现，也有非破坏性实现。

使用 `list.sort()` 或 `list.reverse()` 破坏性地实现。

```python
a = [7, 2, 5, 3, 5, 11, 7]

a.sort()
print(a)
```

```python
a = [2, 3, 5, 7]

a.reverse()
print(a)
```

使用 `sorted(list)` 和 `reversed(list)` 非破坏性地实现。

```python
a = [7, 2, 5, 3, 5, 11, 7]

b = sorted(a)
print(a)
print(b)
```

```python
a = [2, 3, 5, 7]
b = []

for item in reversed(a):
    print(item)
    b.append(item)
print(a)
print(b)
```

# 元组（不可变列表）


元组与列表完全一样，除了它们是不可变的，我们不能更改元组的值。


创建元组：

```python
t = (1, 2, 3)
print(type(t), len(t), t)

a = [1, 2, 3]
t = tuple(a)
print(type(t), len(t), t)
```

元组是不可变的。

```python
t = (1, 2, 3)
t[0] = 42    # 崩溃！
print(t[0])
```

并行（元组）赋值。

```python
x, y = 1, 2
print("x =", x)
print("y =", y)
```

元组对于交换变量非常好用。

```python
x, y = (y, x)
print("x=", x)
print("y=", y)
```

单元素元组语法。

```python
t = (42)
print(type(t), t*5)

t = (42,) # 使用逗号来强制指定类型
print(type(t), t*5)
```

# 列表推导式


列表推导式是一种使用简单的循环在一行内创建列表的简便方法。


普通方式：

```python
a = []
for i in range(10):
    a.append(i)
print(a)
```

简便方式：

```python
a = [i for i in range(10)]
print(a)
```

我们还可以在末尾添加条件（但要保持简单）。

```python
a = [(i*100) for i in range(20) if i%5 == 0]
print(a)
```

# 列表与字符串之间的转换


使用 `list(string)` 将字符串转换为字符列表。

```python
a = list("Datawhale")
print(a)
```

使用 `string_1.split(string_2)` 将字符串转换为由 `string_2` 分隔的字符串列表。

```python
a = "learn python the smart way".split(" ")
print(a)
```

```python
print("/".join(a))
```

```python
a = ["learn", "python", "the", "smart", "way"]
print("".join(a))
print("\n".join(a))
```
